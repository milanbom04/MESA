```python
import mesa_reader as mr
import matplotlib.pyplot as plt
import numpy as np
```

```python
h = mr.MesaData('//wsl.localhost/Ubuntu/home/milan/mesa_workspace/mesa-26.04.1/work_ex1/LOGS_1M/history.data')
age = h.star_age
log_L = h.log_L
log_Teff = h.log_Teff

# Find the index where the star reaches ZAMS (nuclear luminosity >= 99.9% of total luminosity)
Lnuc_div_L = 10**(h.log_Lnuc - log_L)
zams_idx = np.where(Lnuc_div_L >= 0.999)[0][0]
```

```python
# Slice the arrays from zams_idx onward to exclude the pre-main sequence
plt.plot(log_Teff[zams_idx:], log_L[zams_idx:], label='1 $M_\\odot$')
plt.gca().invert_xaxis()
plt.xlabel("Log ($T_{eff}$/K)")
plt.ylabel("log L/$L_{sun}$")
plt.legend()
```
